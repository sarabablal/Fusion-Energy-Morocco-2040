```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/edata.csv", sep=";")
df.head()
```

```python
df.columns
```

```python
raw_2022 = df[df['Year']==2022]
sources = ['Oil', 'Coal and Others', 'Renewables', 'Gas',
       'Nuclear']
values = raw_2022[sources].values[0]
clean_values = []
for v in values:
  v = v.replace("%", "")
  v = float(v)
  clean_values.append(v)
print (clean_values)
```

```python
share = clean_values
colors = ["#555", "#555", "#5792ff", "#555", "#555"]

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(sources, share, color=colors)
ax.bar_label(bars, fmt="%d%%")
ax.set_title("Morocco Total Energy Supply by Source, 2022")
ax.set_xlabel("Energy source")
ax.set_ylabel("Share of total energy supply (%)")
plt.xticks(rotation=10)
plt.tight_layout()
plt.savefig("energy_mix_2022.png")
plt.show()
```
